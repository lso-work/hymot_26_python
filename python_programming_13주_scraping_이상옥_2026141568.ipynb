## 9쪽 - XML 파싱 준비: menu.xml 파일 생성

```python
import xml.etree.ElementTree as et
tree = et.ElementTree(file='menu.xml')
root = tree.getroot()
print(root.tag)
```

```python
menu_xml = """<menu>
    <breakfast hours="7-11">
        <item price="$6.00">breakfast burritos</item>
        <item price="$4.00">pancakes</item>
    </breakfast>
    <lunch hours="11-3">
        <item price="$5.00">hamburger</item>
    </lunch>
    <dinner hours="3-10">
        <item price="$8.00">spaghetti</item>
    </dinner>
</menu>"""

with open("menu.xml", "w", encoding="utf-8") as f:
    f.write(menu_xml)

print("menu.xml 파일 생성 완료")
```

## 9쪽 - ElementTree로 XML 읽기

```python
import xml.etree.ElementTree as et

tree = et.ElementTree(file='menu.xml')
root = tree.getroot()

print(root.tag)
print(type(root))
```

## 10쪽 - XML 내용 프린팅

```python
for child in root:
    print('tag:', child.tag, 'attributes:', child.attrib)
    for grandchild in child:
        print('\ttag:', grandchild.tag, 'attributes:', grandchild.attrib)
```

## 11쪽 - 하위 태그 수 확인

```python
print(len(root))       # menu의 하위 태그 수
print(len(root[0]))    # breakfast의 item 수
print(type(root[0]))
```

## 12쪽 - 특정 태그의 속성 값 접근

```python
print(root[0])                  # root의 첫 번째 자식: Element 'breakfast'
print(root[0].get("hours"))     # get() 함수: 'breakfast'의 'hours' 속성 값 리턴
print(root[0].keys())           # 모든 속성의 key를 리스트로 리턴
print(root[0].items())          # 모든 속성의 key, value 쌍을 리턴
```

## 12쪽 - find method: 태그 이름으로 찾기

```python
lunch = root.find("lunch")

print(lunch)
print(lunch.get("hours"))
print(lunch.keys())
print(lunch.items())
print(type(lunch))
```

## 13쪽 - findall method: 같은 이름의 태그 여러 개 찾기

```python
print(root[0].findall('item'))
```

## 13쪽 - 태그 안의 내용과 속성 출력

```python
print(root[0].findall('item'))

for item in root[0].findall('item'):
    print(item.get("price"))
    print(item.text)
```

## 17쪽 - JSON dumps: 파이썬 딕셔너리를 JSON 문자열로 변환

```python
import json

j1 = {"name": "홍길동", "birth": "0525", "age": 30}

print(j1)
print(json.dumps(j1))
```

## 18쪽 - JSON 문자열 정렬: indent 옵션

```python
print(json.dumps(j1, indent=2))
```

## 18쪽 - 리스트와 튜플도 JSON 문자열로 변경

```python
print(json.dumps([1, 2, 3]))
print(json.dumps((4, 5, 6)))
```

## 19쪽 - JSON loads: JSON 문자열을 파이썬 객체로 변경

```python
j1 = {"name": "홍길동", "birth": "0525", "age": 30}
d1 = json.dumps(j1)

print(d1)
print(json.loads(d1))
print(type(json.loads(d1)))
```

## 20쪽 - JSON file: myinfo.json 파일 생성

```python
myinfo = {
    "name": "홍길동",
    "birth": "0525",
    "age": 30
}

with open("myinfo.json", "w", encoding="utf-8") as f:
    json.dump(myinfo, f, ensure_ascii=False, indent=2)

print("myinfo.json 파일 생성 완료")
```

## 20쪽 - JSON 파일 로드

```python
with open('./myinfo.json', encoding="utf-8") as f:
    data = json.load(f)

print(type(data))
print(data)
```

## 26쪽 - BeautifulSoup으로 HTML 파싱

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen("http://pythonscraping.com/pages/page1.html")
bsObj = BeautifulSoup(html.read(), "html.parser")

print(bsObj.h1)
```

## 27쪽 - BeautifulSoup: 계층 구조 태그 접근

```python
print(bsObj.h1)
print(bsObj.html.body.h1)
print(bsObj.body.h1)
print(bsObj.html.h1)
```

## 27쪽 - BeautifulSoup: prettify로 HTML 구조 출력

```python
print(bsObj.html.prettify())
```
